```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the student dataset
df = pd.read_csv("data/raw/student-mat.csv", sep=";")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
# 1. Dataset dimensions
print("Number of students:", df.shape[0])
print("Number of features:", df.shape[1])

# 2. Column names
print("\nAll columns:")
print(df.columns.tolist())

# 3. Data types and non-null values
print("\nDataset information:")
df.info()
```

```python
# Check missing values
print("Missing values in each column:")
print(df.isnull().sum())

# Total missing values
print("\nTotal missing values:", df.isnull().sum().sum())

# Check duplicate records
print("\nDuplicate records:", df.duplicated().sum())
```

```python
# Statistical summary of grades
print(df[["G1", "G2", "G3"]].describe())

# Distribution of final grades
print("\nFinal grade distribution:")
print(df["G3"].value_counts().sort_index())

# Average final grade
print("\nAverage final grade:", df["G3"].mean())
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["G3"], bins=15, edgecolor="black")

plt.title("Distribution of Students' Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")

plt.show()
```
